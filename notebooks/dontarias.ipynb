```python
import pandas as pd
```

```python
import re
from pathlib import Path

# ==== CONFIGURACIÓN ====
RUTA_CARPETA = r"/Users/alanrobles/Documents/ALAN/GEOSTATS/dontarias/dataet/"
```

```python
carpeta = Path(RUTA_CARPETA)

archivos_por_anio = {}
for archivo in sorted(carpeta.glob("*.xlsx")):
    # ignora archivos temporales de Excel (empiezan con ~$)
    if archivo.name.startswith("~$"):
        continue
    match = re.fullmatch(r"(\d{4})\.xlsx", archivo.name)
    if match:
        anio = int(match.group(1))
        archivos_por_anio[anio] = archivo

if not archivos_por_anio:
    raise FileNotFoundError(f"No se encontraron archivos con nombre AAAA.xlsx en {carpeta}")

print(f"Se encontraron {len(archivos_por_anio)} archivos:")
for anio, ruta in sorted(archivos_por_anio.items()):
    print(f"  {anio} -> {ruta.name}")
```

```python
primer_anio = sorted(archivos_por_anio.keys())[0]
xls_ref = pd.ExcelFile(archivos_por_anio[primer_anio], engine="openpyxl")
nombres_hojas = xls_ref.sheet_names

print(f"Hojas detectadas (tomadas de {primer_anio}.xlsx):")
for h in nombres_hojas:
    print(f"  - {h}")
```

```python
def limpiar_nombre_archivo(nombre: str) -> str:
    """Quita caracteres no válidos en nombres de archivo de Windows/Mac."""
    return re.sub(r'[\\/*?:"<>|]', "_", nombre).strip()


resumen = []

for hoja in nombres_hojas:
    dataframes = []
    for anio, ruta in sorted(archivos_por_anio.items()):
        try:
            df = pd.read_excel(ruta, sheet_name=hoja, engine="openpyxl")
        except ValueError:
            print(f"⚠️  La hoja '{hoja}' no existe en {ruta.name}, se omite ese año.")
            continue

        df.insert(0, "Año", anio)
        dataframes.append(df)

    if not dataframes:
        print(f"⚠️  La hoja '{hoja}' no se encontró en ningún archivo, se omite.")
        continue

    df_concat = pd.concat(dataframes, ignore_index=True)

    nombre_archivo = limpiar_nombre_archivo(hoja) + ".xlsx"
    ruta_salida = carpeta / nombre_archivo
    df_concat.to_excel(ruta_salida, index=False, engine="openpyxl")

    resumen.append((hoja, len(df_concat), ruta_salida.name))
    print(f" '{hoja}' -> {ruta_salida.name}  ({len(df_concat)} filas)")

print("\nListo. Resumen:")
for hoja, filas, archivo in resumen:
    print(f"  {archivo}: {filas} filas")
```
