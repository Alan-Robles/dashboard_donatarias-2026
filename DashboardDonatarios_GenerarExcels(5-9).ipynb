## Librerías

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import plotly.graph_objects as go
import plotly.express as px
```

```python
# Ingresos totales

donativos = pd.read_excel('Ingreso por donativos.xlsx')
relacionados = pd.read_excel('Ingresos relacionados.xlsx')
no_relacionados = pd.read_excel('Ingresos no relacionados.xlsx')

LLAVE = ["Año", "Rfc"]  # ajustar 'ID' a la llave real

# =========================================================
# PREPARACIÓN DE INDICADORES BASE 
# =========================================================
don_efectivo = donativos.groupby(LLAVE)["Monto efectivo"].sum().reset_index()
don_especie = donativos.groupby(LLAVE)["Monto especie"].sum().reset_index()
ing_relacionados = relacionados.groupby(LLAVE)["Monto"].sum().reset_index().rename(columns={"Monto": "Ingresos relacionados"})
ing_no_relacionados = no_relacionados.groupby(LLAVE)["Monto"].sum().reset_index().rename(columns={"Monto": "Ingresos no relacionados"})

total_ingresos = (
    don_efectivo
    .merge(don_especie, on=LLAVE, how="outer")
    .merge(ing_relacionados, on=LLAVE, how="outer")
    .merge(ing_no_relacionados, on=LLAVE, how="outer")
    .fillna(0)
)

total_ingresos["Total Ingresos Anio Fiscal"] = (
    total_ingresos["Monto efectivo"]
    + total_ingresos["Monto especie"]
    + total_ingresos["Ingresos relacionados"]
    + total_ingresos["Ingresos no relacionados"]
)

```

```python
it_df = pd.read_excel("Ingresos totales.xlsx", index_col=0)
```

```python
años = [2020, 2021, 2022, 2023, 2024, 2025, 2026]
```

## 5. Total de fuerza laboral

```python
nomina = pd.read_excel('Nómina.xlsx')

rfc_s = nomina['Rfc'].unique()
```

```python
# Flota laboral
fl = np.zeros((len(rfc_s), 2*len(años)))
```

```python
# Llenar FL con voluntarios y empleados por año y RFC

años = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

for i in range(len(rfc_s)):
    
    fl_i = np.zeros(len(años)*2)
    
    for a in range(len(años)):
        
        
        match = nomina.loc[(nomina['Año'] == años[a]) & (nomina['Rfc'] == rfc_s[i]), ['Plantilla laboral']]
        pl = match.iloc[0, 0] if not match.empty else None
        
        
        match = nomina.loc[(nomina['Año'] == años[a]) & (nomina['Rfc'] == rfc_s[i]), ['Voluntarios']]
        v = match.iloc[0, 0] if not match.empty else None
        
        fl_i[2*a] = v
        fl_i[2*a+1] = pl
    
    fl[i, :] = fl_i

fl_df = pd.DataFrame(data=fl, index=rfc_s, columns=['Voluntarios 2020', 'Empleados 2020', 'Voluntarios 2021', 'Empleados 2021', 'Voluntarios 2022', 'Empleados 2022', 'Voluntarios 2023', 'Empleados 2023', 'Voluntarios 2024', 'Empleados 2024', 'Voluntarios 2025', 'Empleados 2025', 'Voluntarios 2026', 'Empleados 2026'])
```

```python
# IT por año por RFC por año

it = np.zeros((len(rfc_s), len(años)))

for i in range(len(rfc_s)):
    
    it_i = np.zeros(len(años))
    
    for a in range(len(años)):
        
        
        match = total_ingresos.loc[(total_ingresos['Año'] == años[a]) & (total_ingresos['Rfc'] == rfc_s[i]), ['Total Ingresos Anio Fiscal']]
        it_a = match.iloc[0] if not match.empty else None
        
        it_i[a] = it_a
    
    it[i, :] = it_i
```

```python
it_df = pd.DataFrame(data=it, index=rfc_s, columns=años)
it_df.to_excel('Ingresos totales.xlsx')
```

```python
# df con fuerza laboral total por año y RFC

fl_total = np.zeros((len(rfc_s), len(años)))
fl_filtered = np.nan_to_num(fl)

for i in range(len(rfc_s)):
    
    flt_i = np.zeros(len(años))
    
    for a in range(len(años)):
        
        flt_i[a] = fl_filtered[i, 2*a] + fl_filtered[i, 2*a + 1]
        
    fl_total[i, :] = flt_i

fl_total_df = pd.DataFrame(data=fl_total, index=rfc_s, columns=años)
```

```python
fl_df.to_excel('flota_laboral.xlsx')
fl_total_df.to_excel('flota_laboral_total.xlsx')
```

## 6. Costo por empleado (salario promedio)

```python
# Masa salarial por año y RFC 

ms = np.zeros((len(rfc_s), len(años)))

for i in range(len(rfc_s)):
    
    ms_i = np.zeros(len(años))
    
    for a in range(len(años)):
        
        match = nomina.loc[(nomina['Año'] == años[a]) & (nomina['Rfc'] == rfc_s[i])]['Monto salarios']
        ms_i[a] = match.iat[0] if not match.empty else None
        
    ms[i, :] = ms_i

ms_df = pd.DataFrame(data=ms, index=rfc_s, columns=años)
```

```python
# Calcular salario promedio por año por RFC

sp = np.zeros((len(rfc_s), len(años)))

for i in range(len(rfc_s)):
    
    sp_i = np.zeros(len(años))
    
    for a in range(len(años)):
        
        try:
            sp_a = ms[i, a] / fl[i, 2*a + 1]
        except ZeroDivisionError:
            sp_a = 0
        
        sp_i[a] = sp_a
        
    sp[i, :] = sp_i


```

```python
sp_df = pd.DataFrame(data=sp, index=rfc_s, columns=años)
sp_df.to_excel('Salario promedio.xlsx')
```

### 6. Dato General (Salario anual promedio)

```python
spg = np.zeros(len(años))

sp_filtrado = np.nan_to_num(sp)

for a in range(len(años)):
    
    av = np.mean(sp_filtrado[:, a])
    spg[a] = round(av, 2)

spg = spg[np.newaxis, :]

spg_df = pd.DataFrame(data=spg, columns=años)

print(spg_df)
```

## 7. Dependencia del voluntariado

```python
# Calcular dependencia del voluntariado por año por RFC

dv = np.zeros((len(rfc_s), len(años)))

for i in range(len(rfc_s)):
    
    dv_i = np.zeros(len(años))
    
    for a in range(len(años)):
        
        try:
            dv_a = fl[i, 2*a] / fl_total[i, a]
        except ZeroDivisionError:
            dv_a = 0
        
        dv_i[a] = dv_a
        
    dv[i, :] = dv_i
```

```python
dv_df = pd.DataFrame(data=dv, index=rfc_s, columns=años)
```

```python
dv_df.to_excel('dependencia_voluntariado.xlsx')
```

## 8. Costo del órgano de gobierno

```python
# Leer archivo de órgano de gobierno

ogd = pd.read_excel("Órgano de gobierno.xlsx")
```

```python
# Salario de trabajadores gubernamentales

stg = np.zeros((len(rfc_s), 7))

for i in range(len(rfc_s)):
    
    stg_i = np.zeros(len(años))
    
    for a in range(len(años)):
            
        match = ogd.loc[(ogd['Año'] == años[a]) & (ogd['Rfc'] == rfc_s[i])]['Monto salario']
        stg_i[a] = match.sum()
    
    stg[i, :] = stg_i

stg_df = pd.DataFrame(data=stg, index=rfc_s, columns=años)
```

```python
stg_df.to_excel('Salarios trabajadores gubernamentales.xlsx')
```

## 9. Costo por beneficiario

```python
# Leer archivo de destino de donativos

ddd = pd.read_excel("Destino de donativos.xlsx")
```

```python
# Buscar número de beneficiados por año por RFC

ndb = np.zeros((len(rfc_s), len(años)))

for i in range(len(rfc_s)):
    
    ndb_i = np.zeros(len(años))
    
    for a in range(len(años)):
            
        match = ddd.loc[(ddd['Año'] == años[a]) & (ddd['Rfc'] == rfc_s[i])]['Número de beneficiados']
        ndb_i[a] = match.sum()
    
    ndb[i, :] = ndb_i

ndb_df = pd.DataFrame(data=ndb, index=rfc_s, columns=años)
```

```python
ndb_df.to_excel('Número de beneficiados.xlsx')
```
